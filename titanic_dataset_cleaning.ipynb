```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/content/drive/MyDrive/Titanic dataset cleaning/train.csv")
```

```python
print(df.shape)
print(df.columns)
```

```python
print(df['Age'].isnull().sum())
```

```python
age_median = df["Age"].median()
```

```python
df["Age"] = df["Age"].fillna(age_median)
```

```python
print(df['Age'].isnull().sum())
```

```python
df = df.drop('Cabin', axis = 1)
```

```python
print(df.columns)
```

```python
print(df['Embarked'].unique())
```

```python
embarked_mode = df['Embarked'].mode()
```

```python
df['Embarked'].fillna(embarked_mode[0], inplace = True)
```

```python
print(df.isnull().sum())
```
