```python
embarked_mode = df['Embarked'].mode()
```

```python
df['Embarked'].fillna(embarked_mode[0], inplace = True)
```

```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
print(df['Age'].min())
```

```python
df['age_below_lower_bound']= df['Age'] < lower_bound
df['age_above_upper_bound'] = df['Age'] > upper_bound
```

```python
print(df[age_below_lower_bound][['Age','Parch', 'SibSp']])
```
