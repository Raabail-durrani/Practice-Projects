```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/content/drive/MyDrive/Titanic dataset cleaning/train.csv")
```

```python
print(df.shape)
print(df.columns)
```

```python
print(df['Age'].isnull().sum())
```

```python
age_median = df["Age"].median()
```

```python
df["Age"] = df["Age"].fillna(age_median)
```

```python
print(df['Age'].isnull().sum())
```

```python
df = df.drop('Cabin', axis = 1)
```

```python
print(df.columns)
```

```python
print(df['Embarked'].unique())
```

```python
embarked_mode = df['Embarked'].mode()
```

```python
df['Embarked'] = df['Embarked'].fillna(embarked_mode[0])
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())
```

```python
print(df.dtypes)
```

```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
print(df['Age'].min())
```

```python
df['age_below_lower_bound']= df['Age'] < lower_bound
df['age_above_upper_bound'] = df['Age'] > upper_bound
```

```python
print(df[df['age_below_lower_bound']][['Age','Parch', 'SibSp']])
```

```python
print(df[df['age_above_upper_bound']][['Age','Parch','SibSp']])
```
